```python
# Find even numbers

numbers = [10, 15, 20, 25, 30]

for no in numbers:
    if no % 2 == 0:
        print(no)
```

```python
# Find even number

numbers = [10, 15, 20, 25, 30]

def find_even_numbers(number):
    for no in number:
        if no % 2 == 0:
            print(no)

find_even_numbers(numbers)
```

```python
# Calculate cart total using loop

cart = [1200,1500,1000,500,2200]

def find_cart_total(prices):
    
    total = 0

    for price in prices:
        total += price
    return total

total_price = find_cart_total(cart)

print(f"Total price is {total_price}")

```

```python
# Calculate cart total using sum

cart = [100,200,300,400]

def calculate_sum(prices):
    return sum(prices)
    
result = calculate_sum(cart)
print(result)
```

```python
#If total is greater than 2000, give a 10% discount.

prices = [1000, 1000, 500, 500]

def calculation(price):
    return sum(price)
price_sum = calculation(prices)

if price_sum >= 2000:
    after_discount = price_sum-(price_sum * 0.10)
    #after_discount = price_sum * 0.90
    print (after_discount)
else:
    print(price_sum)
```

```python
# Find the Largest Number

numbers = [10, 45, 23, 89, 12]

largest = numbers[0]

for number in numbers:
    if number > largest:
        largest = number
print(largest)
```

```python
# Find the Largest Number

numbers = [10, 45, 23, 89, 12]

def find_largest_value(number):
    largest = number[0]
    for num in number:
        if num > largest:
            largest = num
    return largest

result = find_largest_value(numbers)
print(result)
```
